```python
import numpy
import pandas as pd
from sklearn.datasets import fetch_openml

frequency = fetch_openml(
    data_id=41214,
    target_column=None,
    as_frame=True
).frame

severity = fetch_openml(
    data_id=41215,
    target_column=None,
    as_frame=True
).frame

claims = severity.merge(
    frequency,
    on="IDpol",
    how="left",
    validate="many_to_one"
)
```
